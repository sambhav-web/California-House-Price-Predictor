# Separate input and output

```python
x=df.iloc[:,:-1]
y=df['MedHouseVal']
```

```python
y.shape,x.shape
```

```python
# Train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.linear_model import  LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error,r2_score
```

```python
models={
    'LinearRegressor':LinearRegression(),
    'KNR':KNeighborsRegressor(),
    'SVR':SVR(),
    'XGBOOST':XGBRegressor()
}
scores={}
for i in range(len(list(models))):
  model=list(models.values())[i]
  model.fit(x_train,y_train)
  y_train_pred=model.predict(x_train)
  y_test_pred=model.predict(x_test)
  train_accuracy=r2_score(y_train,y_train_pred)
  test_accuracy=r2_score(y_test,y_test_pred)
  scores[list(models)[i]]=test_accuracy
  print(f'For model {list(models)[i]} :')
  print(f'Train Score : {train_accuracy}')
  print(f'Test Score : {test_accuracy}\n')
```

# Best model is coming out to be XGBOOST with training accuracy as 72 % . Let's perform Hyperparameter Tunning.

```python
from sklearn.model_selection import GridSearchCV
Params={
    'n_estimators':[100,200,300,400,500],
    'max_depth':[1,2,3,4,5]
}
cv_model=[('XGB',XGBRegressor(),Params)]
best={}
for name,model,para in cv_model:
  grid=GridSearchCV(model,param_grid=para,cv=3)
  grid.fit(x_train,y_train)
  best[name]=grid.best_params_
  best['Score']=grid.best_score_
print(best)
```

# Since accuracy is not so great so let's consider simple model

```python
model=XGBRegressor()
model.fit(x_train,y_train)
pred=model.predict(x_test)
```

```python
sns.regplot(x=y_test,y=pred)
plt.title('Actual Vs Predictions')
plt.xlabel('Actual Values')
plt.ylabel('Predictions')
plt.show()
```
